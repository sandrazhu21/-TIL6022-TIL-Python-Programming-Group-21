# Research Objective


Our main goal is to determine the optimal times for electric trucks to charge in the hinterland operations of the Port of Rotterdam, based on dynamic pricing of electricity in order to promote the electrification of the fleet and achieve more sustainable logistics operations.

# Research Questions

How can charging schedules for different truck profiles in the hinterland operations of the Port of Rotterdam be optimized based on dynamic electricity pricing and different charging-mode scenarios?



subquestion 1:

- What do electricity prices look like throughout a day and across the week as well as yearly?

subquestion 2:

- What are the specific battery capacity and energy consumption profiles of short, medium, and bulk truck duties?

subquestion 3:

- Which combinations of truck operating profile and charging modality are feasible?

subquestion 4:

- How do optimized charging schedules reduce costs for each profile compared to night charging?


# Introduction

Truck electrification can play a role in the decarbonization of logistics operations in and around the Port of Rotterdam, but electrification of the truck fleet also depends on operational conditions. For instance, what charging infrastructure can enable trucks to operate through their daily duty? And how do electricity prices affect the choice of different charging infrastructure?

**Operational profiles:**

- short-distance: 0–60 km per one-way trip, normally three round trips per duty
- medium-distance: 60–125 km per one-way trip, normally two round trips per duty
- long-distance: 125–250 km per one-way trip, normally one round trip per duty

**Truck operating characteristics:**

- approximately 60 km/h average speed
- approximately 45 minutes for loading or unloading (dwell time at each origin-destination points)
- battery capacity (to be determined, different for short-medium-long-distance hauls)

**Charging infrastructure modes depending on hour of charging:**

- depot charging only (night time)
- depot plus destination or terminal charging during operational dwell time (night + day time)
- depot, destination and public en-route charging (night + day time + opportunity charging on motorways)

# Data Analysis Pipeline



1.   Data collection: electricity prices (hourly, daily, seasonal); truck specifics (battery capacity, distance and trips, electricity consumption per km, dwell times)
2.   Data cleaning (clean datasets and visualize prices at different time intervals throughout a day/seasonal, energy consumption value per truck profile)
3.   Feature extraction:
          a. create “classes” for each operational profile (truck travel distance, number of trips, battery capacity, minimum state of charge (SOC) to still run, dwell times, operating schedule such as start-end of duty)
          b. create “classes” for charging (kWh, how long charging takes, how battery charge increases, depot/terminal vs opportunity charging in motorway with different charging capacity)
4.   Computation:
          a. create relevant functions such as decreasing battery over trip execution, estimating time needed to complete charging etc.
          b. process each truck profile under different scenarios such as,
                - baseline scenario: truck returns to depot at night and charges overnight
                - depot and destination (dwell time) charging
                - depot, destination and public en-route charging
          c. create a scenario matrix (charging mode vs operational profile) to determine which combinations are feasible (for instance, only depot charging might not be feasible for long-haul)
          d. calculate different pricing profiles for each feasible combination in the matrix
5.    Visualisation: energy costs, scenario matrix, price comparison, if time allows maybe a plotly interactive plot where we can switch across each operational profile and display the feasible charging infrastructure with its associated price
